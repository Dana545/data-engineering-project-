```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


#step one: Examine the initial dataset.
df = pd.read_csv("countries_initial.csv")
print(df.head())    #it will print the first 5 lines so I can match it with the original dataset.
print(df.shape)    #it will print the number of rows and columns of the dataset.
df.info()    #it will show the data types in the dataset and the number of non null values.

#step two: Identify missing values and calculate their percentages.
print(df.isnull().sum())
print(df.isnull().sum() / len(df) * 100)

for column in ["borders", "currencies", "languages", "subregion"]:
    print(column, (df[column] == "None").sum())    #it will count how many values are the text "None" in each of those columns.

for column in ["borders", "currencies", "languages"]:
    print(column, (df[column].astype(str).str.strip(", ") == "").sum())    #it will count the empty values in each column after removing the commas and spaces.

print((df["region"] == "Unknown").sum())    #it will count the countries with an unknown region.
print((df["subregion"] == "Unknown").sum())    #it will count the countries with an unknown subregion.

print((df["area"] == 0).sum())    #it will count the countries with area equal to 0.

#step three: Check data types and identify incorrect ones.
print(df.dtypes)

#step four: Calculate summary statistics for numerical columns:
 #Population
 #Area
 #Other numerical variables
print(df[["population", "area"]].describe())

#step five: Examine categorical distributions, especially countries per region.
print(df["region"].value_counts())
print(df["subregion"].value_counts())

df["region"].value_counts().plot(kind="bar")
plt.title("Countries per region")
plt.xlabel("Region")
plt.ylabel("Number of countries")
plt.show()

#step six: Identify potential outliers using appropriate methods.
for col in df.select_dtypes(include=[np.number]).columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"Outliers in {col}: {outliers.shape[0]}")

```
